## 2. Mismatched Input Shape in CNN
### Question

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, Flatten, Dense

# Generate dummy image data
X_train = np.random.rand(100, 32, 32)
y_train = np.random.randint(0, 10, size=(100,))

# Define CNN model
model = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    Flatten(),
    Dense(10, activation='softmax')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(X_train, y_train, epochs=5)
```

### Answer

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, Flatten, Dense

# FIX: (100, 32, 32) -> (100, 32, 32, 3)
# Explanation: Conv2D needs 4D data (samples, height, width, channels).
# The model expects RGB images (input_shape=(32, 32, 3)), but the data had
# no channel dimension, so the error said "expected min_ndim=4, found ndim=3"
X_train = np.random.rand(100, 32, 32, 3)
y_train = np.random.randint(0, 10, size=(100,))   # 10 classes (0 to 9)

# Define CNN model
model = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    Flatten(),
    Dense(10, activation='softmax')
])

# Explanation: integer labels 0-9 -> sparse_categorical_crossentropy
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(X_train, y_train, epochs=5)
```

### Answer

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.layers import AveragePooling2D, Flatten, Dense, Input
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.models import Model

# Load pre-trained MobileNetV2 model (without its top classification layer)
baseModel = MobileNetV2(weights="imagenet", include_top=False, input_tensor=Input(shape=(224, 224, 3)))

# Add custom layers for binary classification
headModel = baseModel.output
headModel = Flatten(name="flatten")(headModel)
headModel = Dense(128, activation="relu")(headModel)
headModel = Dense(1, activation="sigmoid")(headModel)
model = Model(inputs=baseModel.input, outputs=headModel)

# FIX: define INIT_LR (and EPOCHS) before using them
# Explanation: a variable must be created before it is used.
# INIT_LR was used but never defined, so Python raised NameError
INIT_LR = 0.0001   # small learning rate is good for fine-tuning a pretrained model
EPOCHS = 20

# Optional: learning rate schedule (reduces LR by 4% every 1000 steps)
lr_schedule = tf.keras.optimizers.schedules.ExponentialDecay(
    initial_learning_rate=INIT_LR,
    decay_steps=1000,
    decay_rate=0.96,
    staircase=True
)

# Compile model
# Explanation: the duplicate "opt = ..." line is removed; one optimizer is enough.
# To use the schedule instead, write: Adam(learning_rate=lr_schedule)
opt = Adam(learning_rate=INIT_LR)
model.compile(loss="binary_crossentropy", optimizer=opt, metrics=["accuracy"])
print("Model compiled successfully")
```

## 4. FileNotFound Error
### Question

```python
from PIL import Image

image_path = "flow.jpg"
img = Image.open(image_path)

img.show()
```

### Answer

```python
import os
import numpy as np
from PIL import Image
from tensorflow.keras.preprocessing import image
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

# FIX: "flow.jpg" -> correct file name / path
# Explanation: Python looks for the file in the current working folder.
# "flow.jpg" was a wrong name (or the file was not in that folder),
# so it raised FileNotFoundError. Give the correct name or the full path,
# e.g. r"D:\Hope AI\Quiz Community\Deep Learning\flower.jpg"
image_path = "flower.jpg"

# Explanation: check the file exists first, so we get a clear message
if not os.path.exists(image_path):
    print("File not found. Current folder is:", os.getcwd())
else:
    img = Image.open(image_path).convert("RGB")   # make sure it has 3 channels
    img.show()

    # Prepare the image for MobileNetV2
    img = img.resize((224, 224))                   # size must be a tuple
    img_array = image.img_to_array(img)            # convert image to array
    img_array = np.expand_dims(img_array, axis=0)  # add batch dimension
    img_array = preprocess_input(img_array)        # scale pixels for MobileNetV2

    print("Preprocessed image shape:", img_array.shape)   # (1, 224, 224, 3)
```
